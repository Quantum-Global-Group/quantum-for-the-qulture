````python
import numpy as np
import qiskit
from qiskit import QuantumCircuit
from qiskit.circuit import Parameter
from qiskit.primitives import StatevectorEstimator, StatevectorSampler
from qiskit.quantum_info import SparsePauliOp, Statevector
from qiskit.transpiler import generate_preset_pass_manager

SEED = 7
SHOTS = 512

print("Qiskit version:", qiskit.__version__)
print("NumPy version:", np.__version__)
print("SEED", SEED, "SHOTS", SHOTS)
if qiskit.__version__ != "2.6.2":
    print("This environment is not the tested Qiskit pin. Read the printed results as this environment's results.")
print("Local-only mode. No IBM account has been opened.")


def amplitude_report(circuit):
    """Real amplitudes in Qiskit order: index 0 is 00 and the rightmost bit is qubit 0."""
    labels = ["00", "01", "10", "11"]
    data = Statevector.from_instruction(circuit).data
    parts = []
    for label, value in zip(labels, data):
        value = complex(value)
        if abs(value) < 1e-8:
            continue
        parts.append(f"{value.real:+.4f}|{label}>")
    return " ".join(parts)


def probability_report(circuit):
    raw = Statevector.from_instruction(circuit).probabilities_dict()
    return {str(bit): round(float(prob), 4) for bit, prob in raw.items() if float(prob) > 1e-8}


def count_report(raw):
    return {str(bit): int(count) for bit, count in raw.items()}
````

## Stop 2. Build one circuit

A **quantum circuit** is an ordered list of operations on qubits. This lesson starts with two operations.

- **H** on qubit 0 puts that qubit into an equal superposition of 0 and 1.
- **CX(0, 1)** uses qubit 0 as the control and qubit 1 as the target. The target flips in the piece of the state where the control is 1.

The target state is **Phi Plus**:

```text
|Phi+> = (|00> + |11>) / sqrt(2)
```

A Z-basis measurement of that state returns 00 or 11, each with probability 1/2. The other two strings have probability 0. That is a correlation in one basis. It is not yet a proof of entanglement.

````python
import matplotlib.pyplot as plt


def sample_z_counts(circuit, shots=SHOTS, seed=SEED):
    # Copy so the caller's circuit stays unmeasured.
    measured = circuit.copy()
    # measure_all adds a classical register named meas.
    measured.measure_all()
    # One PUB, written as a one-item list of tuples. shots is a run option.
    job = StatevectorSampler(seed=seed).run([(measured,)], shots=shots)
    raw = job.result()[0].data.meas.get_counts()
    return count_report(raw)


local_counts = sample_z_counts(phi_plus)
print("Phi Plus Z-basis local counts:", local_counts)
print("Total shots:", sum(local_counts.values()))
print("Sampler: StatevectorSampler. This is a local ideal sample, not a device.")

labels = ["00", "01", "10", "11"]
fig, ax = plt.subplots(figsize=(6.2, 3.2))
ax.bar(labels, [local_counts.get(label, 0) for label in labels])
ax.set_title("Phi Plus, local Z-basis counts")
ax.set_xlabel("Bitstring (q1 q0)")
ax.set_ylabel("Shots")
ax.set_ylim(0, SHOTS)
plt.show()
````

## Stop 6. Estimator asks for an average

**Estimator** answers a different question: what is the average value of this observable on the state?

- `ZZ` asks whether the two Z results agree.
- `XX` asks whether the two X results agree.
- Score a measured 0 as the eigenvalue +1 and a measured 1 as the eigenvalue -1. The expectation is the average of those products.
- The expectation is not a bitstring and not a shot count.

Estimator takes the **unmeasured** circuit plus an observable. You do not append Z or X as a gate to ask this question. Putting H into the circuit, as the previous stop did, is a different action: it changes the circuit before a Z measurement.

In a Pauli string, the rightmost character acts on qubit 0, the same side as the rightmost bit. `ZZ` and `XX` put the same letter on both qubits, so that side does not change the value.

One PUB goes in at a time. `result[0].data.evs` is that PUB's expectation. `len(result)` is 1.

````python
# Both flags stay False in the published notebook.
# A job starts only when both are True and the typed confirmation is RUN.
RUN_REAL_HARDWARE = False
I_AUTHORIZE_ONE_QPU_JOB = False
HARDWARE_SHOTS = 256
CHOSEN_BACKEND = ""
hardware_counts = None
hardware_backend_name = None
hardware_job_id = None

print("RUN_REAL_HARDWARE =", RUN_REAL_HARDWARE)
print("I_AUTHORIZE_ONE_QPU_JOB =", I_AUTHORIZE_ONE_QPU_JOB)

if not (RUN_REAL_HARDWARE and I_AUTHORIZE_ONE_QPU_JOB):
    print("Hardware skipped. No IBM account was opened. No job was submitted.")
    print("The local sample above is the result for this session.")
else:
    typed = read_line("Type RUN to submit ONE IBM job, or anything else to cancel: ")
    if typed is None:
        print("No input box is available. No job was submitted.")
        typed = ""
    if typed.strip() != "RUN":
        print("Cancelled. No job was submitted.")
    else:
        try:
            # Imported only after both flags and the typed word. No token is written here.
            from qiskit_ibm_runtime import QiskitRuntimeService
            from qiskit_ibm_runtime.executor_sampler import Sampler as HardwareSampler

            # The service reads the account saved on this machine. Nothing in this cell is a credential.
            service = QiskitRuntimeService()
            if CHOSEN_BACKEND:
                backend = service.backend(CHOSEN_BACKEND)
            else:
                backend = service.least_busy(
                    min_num_qubits=2,
                    operational=True,
                    simulator=False,
                )
            status = backend.status()
            if not getattr(status, "operational", False):
                print("The selected backend is not operational. No job was submitted.")
            else:
                # Measured copy, transpiled for this backend, one Sampler PUB, no observable layout step.
                hardware_circuit = phi_plus.copy()
                hardware_circuit.measure_all()
                hardware_pm = generate_preset_pass_manager(
                    backend=backend,
                    optimization_level=1,
                )
                isa_circuit = hardware_pm.run(hardware_circuit)
                hardware_sampler = HardwareSampler(mode=backend)
                hardware_job = hardware_sampler.run([(isa_circuit,)], shots=HARDWARE_SHOTS)
                hardware_raw = hardware_job.result()[0].data.meas.get_counts()
                hardware_counts = count_report(hardware_raw)
                hardware_backend_name = backend.name
                hardware_job_id = hardware_job.job_id()
                print("Backend:", hardware_backend_name)
                print("Job id:", hardware_job_id)
                print("Hardware shots:", HARDWARE_SHOTS)
                print("Hardware counts:", hardware_counts)
        except Exception as exc:
            hardware_counts = None
            hardware_job_id = None
            print("Hardware path stopped before a usable result.")
            print("Error type:", type(exc).__name__)
            print("No hardware counts were invented. Keep the local sample.")
````
